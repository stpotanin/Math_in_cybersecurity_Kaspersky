Дана таблица данных с именем `dataframe_name`, содержащая информацию о продажах для нескольких продавцов за два года - 2019 и 2020. Необходимо удалить столбец 'sales_2019' из этой таблицы.

Выведите итоговый DataFrame.

```python
import pandas as pd

dataframe_name = pd.DataFrame({'name': ['Alice', 'Bob', 'Charlie', 'David'],
                               'sales_2019': [1000, 2000, 1500, 3000],
                               'sales_2020': [1200, 1800, 1700, 2800]})
dataframe_name.drop('sales_2019', axis=1, inplace=True)
print(dataframe_name)
```
